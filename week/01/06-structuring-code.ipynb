## Conditional Statements (`if`, `else`)

As previously discussed, Python supports comparison statements (`a == b`, `a <= b`, etc.), which return a `bool` value of `True` or `False`.

The most useful application of these statements is to being to introduce conditional statements into your code, which only run if certain conditions are met:

```python
a = 34
b = 12

if a * b > 400:
    print("The product of a and b is greater than 400.")

print("This statement will appear regardless of the if statement above, as it is not indented.")
```


Note how the code is structured: a colon indicates the beginning of the statement code block, and anything the block is indented by four spaces.

You can introduce more optionality by using the `else` statement:

```python
c = 5445

if type(c) is str:
    print("Let's repeat", c, "five times:", c * 5)
else:
    print("c is not a string! I only want to repeat strings.")
```

Or even an intermediate `elif` (else-if) statement:

```python
d = 304

if d > 400:
    print(d, "is greater than 400.")
elif d > 300:
    print(d, "is greater than 300.")
elif d > 200:
    print(d, "is greater than 200.")
elif d > 100:
    print(d, "is greater than 100.")
else:
    print(d, "is less than or equal to 100.")
```

You must begin with `if` and and with `else`, and all `elifs` must be inbetween. `else` does not get to have a condition: it is what happens if all other checks fail.

### Logical operators (`and`, `or`)

You might have noticed an issue with the above loop: it must be constructed in the correct way. If we invert the order, then `d` will trigger the first catch (`> 100`), even if it is more suited for a later type (`> 300`):

```python
d = 304

if d > 100:
    print(d, "is greater than 100.")
elif d > 200:
    print(d, "is greater than 200.")
elif d > 300:
    print(d, "is greater than 300.")
elif d > 400:
    print(d, "is greater than 400.")
else:
    print(d, "is less than or equal to 100.")
```

We should be probably be checking if this number falls _between_ certain values. This requires muliple checks. We can do this using logical operators, such as `and`, `or`, or `not`. Here is an example of combining these values, so that `e = 304` will not trigger the first `e > 100` check:

```python
e = 304

if e >= 100 and e < 200:
    print(e, "is greater or equal than 100 and less than 200.")
elif e >= 200 and e < 300:
    print(e, "is greater or equal than 200 and less than 300.")
elif e >= 300 and e < 400:
    print(e, "is greater or equal than 300 and less than 400.")
elif e >= 400:
    print(e, "is greater or equal than 400.")
else:
    print(e, "is less than 100.")
```

Here is a more complex check, which will confirm that a variable is a number before doing something with it. 

```python
f = "three hundred and four"

if type(f) is float or type(f) is int:
    print(f, "is a number.")
else:
    print(f, "is not a suitable Python number type. Please use a numeric value.")
```

This is a useful check to make in your code, especially if you are creating generic functions that users could input arbrirtrary values into. You can combine type checking with "errors", which will break the code and alert the user that something has gone wrong. Here, we '`raise`' a '`TypeError`', which a short message letting the user know what has gone wrong.

```python
# Example of type checking and raising an error

if type(f) is not float and type(f) is not int:
    raise TypeError(f, "is not a suitable Python number type. Please use a numeric value.")
```

There are a whole range of potential errors (e.g. `TypeError`, `ValueError`), but these are just for information - in truth, any will work, and they are useful for informing the user what might have gone wrong even if they aren't being used perfectly under the correct heading.

## Loops (`for`, `while`)

### `for` loops

Let's return to our list of days of the week:

```python
weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

weekdays
```

If we wanted to print these out line-by-line, we could certainly do so, in an inefficient way:

```python
print(weekdays[0])
print(weekdays[1])
print(weekdays[2])
print(weekdays[3])
print(weekdays[4])
print(weekdays[5])
print(weekdays[6])
```

But that's not very efficient - eventually, we will want to not just print the days of the week but perform operations hundreds or thousands of data points. And what if we don't know what the length of the list is?

Instead, we can use a `for` loop:

```python
for day in weekdays:
    print(day)
```

Note that this creates a new variable, `day`, that changes as the list is looped through. This a shorter and more flexible way of moving through long lists.


The "loop" is defined by the text after the `for x in y:` line that is _indented_ by a number of spaces. Python will try and accept a number of different indentation lengths, but the standard is four spaces.

```python
day_number = 0

print("Let's begin counting the days of the week!")

for day in weekdays:

    # The code block is all indented by four.

    day_number += 1  # This is the same as day_number = day_number + 1

    print("Day", day_number, "is", day)

# Now, after all the weekdays are looped through, the code will carry on.

print("Those are all the days of the week!")
```

Note that after the loop is finished, the looping variable (here, `day`) will persist, stuck on the last loop. Be careful of this!

```python
print("The day of the week is", day)
```

You can use a `for` in combination with other functions to create operations of set lengths, such as `range`:

```python
print("Let's count to five!")

for i in range(5):
    print(i)

print("Whoops, I forgot Python starts counting at 0!")

for i in range(5):
    print(i + 1)
```

We already know the `len` function: we can combine this with `range` to give a length based on the list length:

```python
for i in range(len(weekdays)):
    print(weekdays[i], "is at index value", i, "and is day", i + 1, "of the week.")
```

You can also loop through two (identical length) lists together using the `zip` function:

```python
countries = ["United Kingdom", "France", "Germany", "Italy", "Spain"]
capitals = ["London", "Paris", "Berlin", "Rome", "Madrid"]

for country, capital in zip(countries, capitals):
    print("The capital of", country, "is", capital)
```

Or see how we can merge them with dictionaries to do something even more interesting:

```python
# Look, we can create dictionaries of dictionaries! This is a more complex data structure that allows us to store more information about each country.
# We could find the capital of france, for instance, by using countries_dict["France"]["capital"].
countries_dict = {
    "United Kingdom": {"capital": "London", "currency": "Pound Sterling"},
    "France": {"capital": "Paris", "currency": "Euro"},
    "Germany": {"capital": "Berlin", "currency": "Euro"},
    "Italy": {"capital": "Rome", "currency": "Euro"},
    "Spain": {"capital": "Madrid", "currency": "Euro"}
}

for i in range(len(countries)):
    print("The capital of", countries[i], "is", countries_dict[countries[i]]["capital"], "and the currency is", countries_dict[countries[i]]["currency"])
```

### `while` statements

A cousin of the `for` statemnets is `while` statements. The `while` loop will continue to execute as long as a given condition is true:

```python
i = 0

while i <= 7:
    print(i)
    i += 1

print("The while loop has finished, and the value of i is now", i)
```

Be careful - if, for whatever reason, your condition never is met, the loop will run forever!

### `break` and `continue` statements

There are two other statements we can use to refine our `for` and `while` loops for exceptional circumstances, by combining with the `if` statement. `break` will immediately end the loop if a condition is met, even if the loop isn't finished:

```python
i = 0

for i in range(10):

    print(i)
    
    if i == 4:
        print("The loop will now break, as i is equal to 4.")
        break
```

Meanwhile, a `continue` statement can be combined with an `if` statement to skip the rest of the code in the loop and start a new one:

```python
i = 0 

for i in range(10):

    # if number is even, skip to the next iteration of the loop without printing
    if i % 2 == 0:
        continue

    # else, the value will be printed
    print(i)
```

## Functions

### Introduction to Functions

We have begun to see how we can use conditional statements and loops to construct more complex code. However, if we're writing complex code, it would be nice to be able to use it again and again without needing to rewrite or copy and paste code!

This is where **functions** come in. Functions are a block or organised, reusable code that act as self-contained programmes for performing a specific task, which you can use again and again throughout your code. One of the basic principles of good programming is 'don't repeat yourself': any time you find yourself duplicating lines of code in your programmes, consider whether you should be wrapping that code into a function.

We have already used many functions so far, such as `print()` or `range()`. Functions are identifiable by the brackets after them: these brackets indicate that a function will accept **arguments**.

Let's create a simple function to show what a funciton looks like behind the scenes:

```python
def farenheit_to_celsius(temp_f):
    temp_c = (temp_f - 32) * 5 / 9
    return temp_c
```

In the block above, the `def` statement tells Python that we are about to define a function. `farenheit_to_celsius` is the name of our funciton, and the `f` in brackets shows that it will accept one argument, the temperature in farenheit. The code block performs an operation on this argument, and then is told to `return` the new argument `temp_c`.

Let's now try and use this a few times:

```python
farenheit_to_celsius(32)  # note that the `temp_c` variable is not accessible outside of the function: it is simply returned as an output
```

```python
farenheit = 100

celscius = farenheit_to_celsius(farenheit)  # we can set the output to any named variable to keep it

print(farenheit, "degrees Farenheit is equal to", celscius, "degrees Celsius.")
```

### Functions with no parameters

You aren't just limited to one parameter. Sometimes, you might not want to `return` anything at all -- everything that needs to happen, happens inside the function. However, you can consider a function it's own little programme: you won't be able to access any variables that are not returned. See this example:

```python
def greeting(name):
    secret_number = 5  # this variable is only accessible within the function, and will not be accessible outside of it
    print("Hello", name, "!")
```

```python
greeting("Alice")
```

Even though we have run the function, the `secret_number` variable only exists inside of it: is is not retained outside, in the "global namespace".

```python
print(secret_number)
```

However, the inverse is not true: functions can access the global namespace:

```python
temperature = 30  # a random temperature in degrees Celsius, existing in the global namespace

def kelvin_to_celsius_incorrect(temp):
    temp_c = temperature - 273.15
    return temp_c

kelvin_to_celsius_incorrect(273.15)
```

Note that we have sloppy with our variables here - even though we have the parameter `temp` as an argument, we used the variable `temperature` in the code block. Hence, our output is incorrect! These are things you must be wary of when writing functions.

### Functions with more than one parameter

It is possible to have functions with more than one argument:

```python
def is_it_too_hot(temp_c, max_temp_c):
    if temp_c > max_temp_c:
        print("Yes, it is too hot!")
    else:
        print("No, it is not too hot.")
```

```python
is_it_too_hot(30, 25)
```

Note that you must provide your arguments to the function in the exact order specified by the function parameters (e.g. in this case, `temp_c` followed by `max_temp_c`).

### Functions with optional parameters

Sometimes, you might want to provide a number of possible parameters within a function, but keep some as 'default' parameters that do not need to be passsed. Here is an example, which sets some default values that do not need to be provided every time:

```python
def comfy_temperature(temp_c, min_temp_c=10.0, max_temp_c=25.0):

    if temp_c < min_temp_c:
        print("It is too cold!")
    elif temp_c > max_temp_c:
        print("It is too hot!")
    else:
        print("It is a comfy temperature.")
```

Now, we don't need to set the `min_temp_c` or `max_temp_c`:

```python
comfy_temperature(28)
```

However, if you disagree with any of the default values, you can change them:

```python
comfy_temperature(28, max_temp_c=30.0)
```

### Multiple return values

Functions can also return multiple values, which you must understand how to return:

```python
def kelvin_to_all(temp_k):
    temp_c = temp_k - 273.15
    temp_f = (temp_c * 9 / 5) + 32
    return temp_c, temp_f
```

If we try and accept this output to a single item, we will see we get a tuple of multiple values:

```python
output = kelvin_to_all(300)

output
```

However, can directly 'unpack' the tuple values when calling the function, by providing the same number of variables for assignment as is returned by the function:

```python
temperature_c, temperature_f = kelvin_to_all(300)

print("300 Kelvin is equal to", temperature_c, "degrees Celsius and", temperature_f, "degrees Farenheit.")
```

If we are not even interested in one of the outputs, it is convention to 'ignore' it by assigning it to a single underscore, `_`:

```python
temperature_c, _ = kelvin_to_all(300)

print("Who cares about Farenheit? 300 Kelvin is equal to", temperature_c, "degrees Celsius.")
```

### Documenting a function

Functions are no cleverer than the code we originally wrote: we can do things like give it too many arguments or the wrong arguments, and raise an error:

```python
farenheit_to_celsius("two hundred")
```

```python
farenheit_to_celsius(203, 234, 344)
```

"Good code" will include additional effort, such as "docstrings" (defined within triple quotation marks) that describe a function to a user, and type checking, so that users are warned when the wrong function is being used. Here is an example of a well-documented version of the function:

```python
def farenheit_to_celsius_documented(fahrenheit):
    """
    Converts a temperature in degrees Farenheit to degrees Celsius.

    Parameters
    ----------
    fahrenheit : float or int
        The temperature in degrees Farenheit to be converted to degrees Celsius.
    
    Returns
    -------
    celsius : float
        The temperature in degrees Celsius.

    Raises
    ------
    TypeError
        If the input is not a float or int, a TypeError will be raised.
    """

    # type checking to ensure the input is a float or int
    if type(fahrenheit) is not float and type(fahrenheit) is not int:
        raise TypeError(fahrenheit, "is not a suitable Python number type. Please use a numeric value.")

    celsius = (fahrenheit - 32) * 5 / 9
    
    return celsius
```

Docstrings can be consulted using `help(function_name)`. For instance:

```python
help(farenheit_to_celsius_documented)
```

Let's see what happens if we try and break it:

```python
farenheit_to_celsius_documented("hello")
```

## Loading additional modules

So, functions are a useful way of wrapping our data and begin to do cool things with our code! We can write them ourselves, or use some of the simple ones that have been provided (e.g. `print`, `range`, etc.)

However, we have access to a much wider range of functions than those that are simply built into Python. We installed `conda` earlier so we could begin to access some of these tools. Functions can be grouped into larger groups known as "modules" or "packages", often used interchangeably (and also called "libraries" in other programming langauges). We mentioned that there are pre-installed packages known as the "standard library" that provides additional functionality.

The most classic example is the `math` library. We can summon these libraries into our namespace using the `import` function:


```python
import math
```

`math` has a number of accessible functions. You can use them by using a dot (`.`), and then the name of the function you wish to use. The dot is an 'attribute access operator' - it tells Python to look _inside_ the `math` object to find what you want to use. `math` has some useful functions:

```python
math.sqrt(16)  # Square root
```

```python
math.pi  # `pi` is not a function per se - it doesn't have any parentheses after it to accept arguments.
         # Instead, it is a constant value that can be called to be used in calculations.
         # After all, who wants to remember pi to so many significant figures?
```

```python
math.sin(math.pi / 2)  # Sine of 90 degrees (or pi/2 radians)
```

You can also import specific functions from packages rather than the whole package:

```python
from math import sin

sin(2.5)
```

You could theoretically import all the functions using a wildcard operator: `from math import *`. This would give you access to everything (`sin`, `sqrt`, `pi`, etc. etc.) without needing to specify it came from `math`. However, this is generally seen as a bad idea - how do you know what you might be overwriting, and how do you keep track of where things come from?

You can also choose to "rename" modules or functions on import using `as` - for example, `import math as m` to get `m.sin()`.

You can make use of this import functionality to hide away functions you don't like. For instance, by putting all your functions in a file `mytools.py` script, located within the same directory as the Jupyter Notebook you are using, you could `import mytools`. It is also possible to install them anywhere by creating your own packages, but that is for another course!
