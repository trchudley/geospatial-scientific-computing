### Assigning variables

What is more useful is being able to 'remember' outputs across multiple commands. We can do this by assigning a variable to a specific name - for instance:

```python
x = 5

y = 5 * 2

x * y
```

Variables can be reassigned to other values after they are set:

They are case sensitive -- so you must be careful here

Traditionally, multi-word variables in Python are written in 'snake case', and seperated by underscores (e.g. `my_variable`). Some other languages (e.g. MATLAB) may utilise camel case, where the first word is lower case and every new word is a capital (`myVariable`).

One important thing to remember in Jupyter Notebooks is that variables are remembered and updated in the order you ran the cells. For instance, take this example:

```python
temperature_celcius = 20
```

```python
temperature_fahrenheit = (temperature_celcius * 9/5) + 32

temperature_fahrenheit
```

Now, if we were to run the below cell, and then run the above cell again, the value would change as `temperature_celcius` now refers to `18`, not `20`. Make sure to not make mistakes by running things out of order and losing track!

```python
temperature_celcius = 18
```

Another common catch is that, so far, we've been tracking the output of our operations by what is printed below the cell when we run it. However, only the final operation of the cell is printed -- for instance:

```python
10 / 2  # Will not be printed

20 / 2  # Will not be printed

30 / 2  # Last line of the cell: will be printed
```

Additionally, some operations don't always have an output - for instance, when assigning the output to a variable:

```python
my_variable = 20 / 2  # Assigned to a variable, not printed
```

You can use the `print` function to force Python to print output at any stage in the cell, not just the last.

```python
print(10 / 2)

my_variable = 20 / 2

print(my_variable)

30 / 2  # Will be printed regardless as is the last operation in the cell
```

### Numbers (`float`)

You might have noticed above that whilst most outputs return an integer without a decimal (e.g. `4`) whilst the division operations return a decimal number (e.g. `4.0`). 

This is because Python actually has two different "data types" for numbers: integers (`int`) and floating-point (`float`; i.e. decimals). You can mix and match these in basic maths and Python won't be confused, but any operation where at least one of the inputs is a `float` will return a float in kind:

```python
3 - 1  # Both are integers, so the result is an integer
```

```python
3 - 1.0  # 1.0 is a float, so the result is a float
```

Division will always return a float, even if the output could be an integer.

```python
5 / 2
```

```python
4 / 2  # Could be an integer output, but Python will return a float
```

The only exception is using the special floor division operator (`//`), which returns the quotient of the division without the remainder. This will return an integer if both inputs are also integers.

```python
5 // 2
```

(the inverse of the floor division operator is the module operator, `%`, by the way, which also obeys this `int` rule)

```python
quotient = 5 // 2
remainder = 5 % 2

print("The quotient of 5 divided by 2 is", quotient, "and the remainder is", remainder)
```

Unlike a lot of other languages, Python can deal with a lot data type management in the background without you even noticing. In lower-level languages, you might to explicitly declare a data type before using one, which is a lot more arduous! However, there is a downside to this: sometimes, the things Python can be doing in the background may be not ideal, or against your will.

Take `int` and `float`, for example. Computers will store all values as binary. `int` values are stored 'exactly', as all integer values can be represented in binary form (e.g. `232` is `11101000`). However, `float` values can only be stored to a 64-bit precision (~15-17 significant digits). As the maths is being done in binary but presented base-10, in very rare events you can have notice some unusual things happening when doing floating-point maths!

```python
0.1 + 0.2
```

Of course, when you're running float operations at scale, an error of the order of 10<sup>-18</sup> isn't really of any importance for most of our purposes. But bear this in mind - sometimes, when computers "just do stuff for you", what is happening in the background might not be what you intended. This is as true for a calculator as it is for AI!

If you're ever unsure what object a Python variable is, you can check using the `type()` function:

```python
a = 10

type(a)
```

```python
b = 10.0

type(b)
```

### String (`str`)

Let's introduce a new data type: the string, which we use to represent text. Strings are identified by Python as being between quotation marks (either of `'this'` or `"this"`).

```python
name = "John"

name
```

```python
type(name)
```

Strings have different properties to numbers: operations produce different outputs. Some of them make sense - for instance, it seems reasonable that "adding" strings together would concatenate them into a single object:

```python
"My name is " + name
```

Strings also have a number of special functions - see here for a few examples

```python
"the upper case function".upper()
```

```python
"The Lower Case Function".lower()
```

```python
"the split function splits a string based on a single character".split(" ")  # split based on a space
```

This last function produces a new data type - a `list` - which we will discuss below.

### Data type 'catches'

Some operations no longer have any meaning, and will return an error. For instance, how do you 'divide' text?

```python
"My name is " / name
```

The answer is: you don't! This raises a📈 `TypeError`, which means that Python can't make sense of the data types you've put into a requested operation (here, it reports `unsupported operand type(s) for /: 'str' and 'str'
`). You must be aware of the data types of your variables, and take steps to ensure your code remains compatible with its inputs.

As another example, it also doesn't make a lot sense to add together a number and a string. Note how the below fails - we have used quotation marks, so the second variable is evaluated as a string!

```python
c = 1
d = "2"

c + d
```

One way we can ensure that different data types work together is to convert between data types explicitly. For instance, `int("5.0")` would return `5.0`, and `str(5.0)` would return `"5.0"`:

```python
d_fixed = int(d)

c + d_fixed
```

However, there are limits to this - for instance, `int` and `float` can only create numbers from strings that represent numerical values. This, for instance, will fail:

```python
int("letters")
```

For presenting feedback to the user, the `print` function will try and turn everything into a string and present it to the user. 

So whilst this doesn't work, because you can't "add" a `str` and an `int`:

```python
"Here is the number two: " + 2
```

This will, as `print` will just try and mash everything together.

```python
print("Here is the number two: ", 2)
```

Note that, in the `print` function, the variables we want to concatenate are seperated by commas rather than `+` signs.

Another, more modern, way of placing everything into a string is to use "f-strings". By putting the letter "f" in front of a string, Python will attempt to resolve everything between curly brackets (`{}`) and output a string.

```python
# example of f-strings

name = "John"

print(f"My name is {name}.")

print(f"We can also do math in f-strings: 2 + 2 = {2 + 2}")
```

There's more fun stuff we can do with f-strings, but we will leave it here for now...

### Booleans (`bool`)

Another valuable base data type is the Boolean, which evaluates to either `True` or `False`. These will become invaluable later, when we beging to introduce decisions and logic into our code. For now, let's show how we can use comparison operations to produce `True` or `False` outcomes.

```python
5 > 10  # Greater than
```

```python
10 >= 10  # Greater than or equal to
```

```python
0.1 == (1 / 10)  # Equal to; the operation will evaluate both sides before checking if they are the same
```

```python
"FizzBuzz" == ("Fizz" + "Buzz")  # This will work for any data type, not just numbers.
```

```python
5 != "five"  # Not equal to
```

```python
type(1) is int  # "is" is a special case of equality checking -- more on that below -- but is used for checking data types
```

```python
(10 == 10) is True  # or checking if something is True or False
```

```python
type(1) is not float  # "is not" is the opposite of "is" and checks if two things are not the same
```

This table summarised potential comparisons:

| Operation | Meaning |
| - | - |
| `<` | Less than |
| `<=` | Less than or equal to |
| `>` | Greater than |
| `>=` | Greater than or equal to |
| `==` | Equal to |
| `!=` | Not equal to |
| `is` | Object identity |
| `is not` |  Negated object identity |

Note the similar, but slightly different, use of the `==` and `is` comparisons: `==` is for "value equality", when you would like to know if two variables have the same value. `is` si for "reference equality", when you would like to know if two variables refer to the same thing.
