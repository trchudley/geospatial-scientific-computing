# A Python Primer

The rest of this week aims to give you a very brief overview of the Python syntax, designed for those with little prior experience in Python. It will form the foundations of how we interact with Python for the rest of the course. 

If you familiar with Python, then most of this will not surprise you, although it is worth going through once to make sure. However, if you are not relatively new to Python, then it is critical you take the time to understand the basics of what's going on, and you might need to spend a little more of your time outside of the lab class doing some supplementary Python prep - a few example resources are shown on the home page for this week, the best of which is the excellent [Software Carpentry "Programming with Python" course](https://swcarpentry.github.io/python-novice-inflammation/).

Regardless of whether you are familiar with Python or whether you have never typed a line of code in your life, the most important thing to achieve in our lab time this week is to make sure we have succesfully installed Conda, Python, and Jupyter, and are comfortable loading and interacting with Jupyter Notebooks. 

## Opening a new notebook

As described on the [previous page](03-installing-jupyter.md), open up your command prompt/terminal and ensure you are in the right directory (you might want to make a new one in a sensible place - e.g. `Documents/MSc/GEOMG0081/weekly_notebooks`). Then, run your environment Jupyter Lab (run `conda activate gsc`; and then type `jupyter lab`).

If you are happy you are in the right directory (navigate around using the left-hand sidebar), then you can create a new notebook by clicking a 'new notebook' button in the middle of the page. Remember to call it something sensible - by default, it will be called `Untitled.ipynb` (all Jupyter Notebook end `.ipynb`, a vestigial callback to their original name: "iPython Notebooks").

As you go through these pages, you are welcome to use the Jupyter Notebook to make notes (both text and code), write code and outputs, and play with changes and modifications to see what happens. You can create a new Jupyter Notebook, or you can open with / edit / play with this one, by clicking the 'Download' symbol above and to the right of the page title. Download it and ensure it is saved in a sensible location within your Desktop, and one you can access from your Jupyter Lab environment.
