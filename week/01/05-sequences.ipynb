# Python Sequences

In the previous section, we learned about basic Python data types and how we can assign them to variables and play with them. However, there are more data types that allow us to chain lots of individual items together.

### Lists (`list`)

Using a `list` data type, we can store many values together in a single variable. Variabels are stored between square brackets, seperated by commas:

```python
cities = ["Bristol", "London", "Manchester", "Edinburgh", "Cardiff", "Newcastle", "Liverpool", "Birmingham", "Leeds", "Glasgow"]

cities
```

```python
type(cities)
```

Lists have their own set of tricks - for instance, the `len` funcion will tell you how long the list is:

```python
len(cities)
```

And you can do additional checks - such as testing whether an item is `in` a list.

```python
"Edinburgh" in cities
```

The list also plays nice with other functions, such as `sorted`, which, for a list of strings, will return the items sorted alphabetically:

```python
sorted(cities)
```

Note that lists can store a number of abritrary data types:

```python
my_list = ["pi", 3.14, True]
```

But that this will cause certain operations, such as `sorted`, to fail.

```python
sorted(my_list)
```

### Indexing

To access an individual value in a list we need to use an index value. An index value is a number that refers to a given position in the list. Let’s check out the first value in our list as an example by printing out cities[1]:

```python
cities[1]
```

Wait, what? This is the second item. Confusingly, Python (and many other programming languages) start values stored in collections with the index value 0. Thus, to get the value for the first item in a list, we must use index `0` (and the second item, `1`, and so on). Let do this:

```python
cities[0]
```

You can also count 'backwards' through the list. `-1` refers to the final position, `-2` to the second-to-last, and so on.

```python
cities[-2]
```

If you ask for an index position that doesn't exist, you will receive an error:

```python
cities[56]
```

You can, however, select a range of values - here, we are selecting between the first and third index position (i.e. from the second item up to, but including, the fourth):

```python
cities[1:3]
```

### Other list functions

Note that lists also have their own special functions to do things, so you don't always have to be fiddling with indexing

```python
another_list = ["one", "two", "three"]
print(another_list)
```

```python
another_list.append("four")  # Add a new item onto the end of the list
print(another_list)
```

```python
another_list.remove("two")  # Remove a specific item from the list
print(another_list)
```

```python
another_list.pop(2)  # Remove the item at index position 2 (the third item)
print(another_list)
```

```python
another_list.clear()  # Remove all items from the list
print(another_list)
```

### Mutability (`tuple`)

Lists are "mutable", meaning the can be changed. We can add things to the list:

```python
cities[2] = "Plymouth"

cities
```

The _mutable_ `list` datatype has an _immutable_ cousin, called the `tuple`. These cannot be changed! You may want to use a tuple when you want to store a collection of items that should not be changed, for example the days of the week or the months of the year.

```python
weekdays = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")

weekdays[-2] = "Three day weekend!"  # This will fail because tuples are immutable, meaning their contents cannot be changed after they are created.
```

### "Mappable" objects (`dict`)

It's all very well calling individual items in a list using the index number (e.g. `cities[4]`), but sometimes you will want to extract specific information without remembering an index value. "Mappable" objects, of which the basic Python example is a Dictionary (`dict`), allow valuables to be retrieved from "key-pair" values. For instance: 

```python
student = {
    "name": "John",
    "age": 23,
    "city": "Bristol"
}
```

Now, specific instances from your item can be recalled from the "key":

```python
student["name"]
```

```python
print(student["name"], "is", student["age"], "years old and lives in", student["city"])
```

You can recall what the keys are using the `keys()` function:

```python
student.keys()
```

Dictionaries are mutable, so you can add more information as you go:

```python
student["favourite_food"] = "Pizza"

student
```

Boths keys and pairs in a dictionary can be any data type, although it is generally recommended you stick to `str` values for keys!

```python
# Don't do this!
new_dict = {
    133: ["apple", "banana", "cherry"],
    True: 125,
}

new_dict[True]
```

However, one particularly ueful use case is that we can have dictionaries containing dictionaries. 

```python
countries_dict = {
    "United Kingdom": {"capital": "London", "currency": "Pound Sterling"},
    "France": {"capital": "Paris", "currency": "Euro"},
    "Germany": {"capital": "Berlin", "currency": "Euro"},
    "Italy": {"capital": "Rome", "currency": "Euro"},
    "Spain": {"capital": "Madrid", "currency": "Euro"}
}
```

```python
countries_dict["Spain"]
```

```python
countries_dict["Spain"]["capital"]
```
