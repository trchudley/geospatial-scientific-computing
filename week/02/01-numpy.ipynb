## What's wrong with built-in functionality?

 The package we are examining in this section -- `numpy` -- is the foundation of everything else we built in scientific python. It allows us to do 'proper maths'.

 But wait -- didn't we do maths last week? We even imported a `math` module that allowed us to do more advanced things!

````python
import math
````

Indeed, `math` has [quite an extensive library of functions](https://docs.python.org/3/library/math.html) that we can explore:

````python
# Calculate e raised to the power of 5

result = math.exp(5)
print(f"e^5 = {result}")
````

````python
# Convert from degrees to radians

degrees = 45
radians = math.radians(degrees)

print(f"{degrees} degrees is equal to {radians} radians")
````

````python
# calculate the cosine of 60 degrees

cosine_60 = math.cos(math.radians(60))

print(f"The cosine of 60 degrees is {cosine_60}")
````

The main problem is that the standard data types available to us aren't naturally good at doing large mathematical operations. Let's, for example, imagine we have some temperatures in degrees celcius (in a `list` form), and want to convert them to Kelvin:

````python
temperatures_c = [18.2, 19.1, 20.5, 21.0]
````

````python
temperatures_k = temperatures_c + 273.15
````

Uh oh. Python has thrown an error! It doesn't care what's inside the list (be they strings, integers, dictionaries, etc...), so it doesn't know what we mean when we try and "add one to a list". For instance, it could be an equally valid interpretation that we want to append the value `1` to the end of the list, for instance (equal to `temperatures_c.append(1)`).

Okay, let's take another route. If the `list` capability isn't sure what we mean when we add a single number, why not create a list of equal length, all of `273.15`. Then, we will be able to add each element of one list to each element of another. This makes sense, right?

````python
add_kelvin = [273.15, 273.15, 273.15, 273.15]

temperatures_k = temperatures_c + add_kelvin

temperatures_k
````

Oh dear, this still isn't what we want! Python thinks that "adding" lists together means _concatenating_ lists, because lists are designed to handle all types of objects, not just numbers.

In fact, the easiest way to perform this is probably to loop through every temperature individually, adding values:

````python
temperatures_k = []

for t in temperatures_c:
    temperatures_k.append(t + 273.15)

temperatures_k
````

This isn't convenient at all for such a simple operation! Additionally, this is _expensive_: we all have very advanced computers nowadays, but in this loop format, our computer will try and solve these operations one-by-one. We can use a special Notebook function called `%%time` to show us how long each cell takes to run. Let's try and calculate a square root a large amount of times - say, 50 million.

````python
%%time

n_sqrt = []

for n in range(50_000_000):  # Underscores in numbers are just for readability, they don't affect the value
    n_sqrt.append(math.sqrt(n))
````

On my laptop, it takes 1.89 seconds to calculate a square root 50 million times. This might seem impressive, but, as we will show later, it is trivial compared to what modern computing is capable of!

We need a tool that allows us to operate more easily on large numerical datasets.


## The ease of `numpy`

Let's start using `numpy` instead of built-in Python and `math`.

````python
import numpy as np
````

`numpy`, conventionally shortened to `np` when importing, introduces another type of object: the numpy array.

````python
temperatures_c = np.array([18.2, 19.1, 20.5, 21.0])

temperatures_c
````

Unlike lists, which happily accept all data types, numpy arrays accept numbers, and only numbers. What's more, they have to be of a single type - you can't mix integers and floats, as we were doing before.

This might seem like it things less flexible than lists. However, because we know that everything inside an array is numerical, `numpy` can accept what instict said would be a 'sensible' way approaching operations from before. For instance, we can add a single value to the whole list:

````python
temperatures_c + 273.15
````

It will also allow you to add two arrays of equal size, so all our approaches will work!

````python
add_kelvin = np.array([273.15, 273.15, 273.15, 273.15])

temperatures_c + add_kelvin
````

Another key factor to understand is that numpy is _vectorising_ our operations. This means we are computing entire sets of values at once, rather than explitly _looping through these lists one at a time_ like we were doing before.

We can see the difference when doing the same thing as we were doing before: calculating the square root of 50 million numbers. This time, we use custom `numpy` functions: `np.arange()` to generate a numpy array rather than `range`; and `np.sqrt` to apply a square root to a whole numpy array, rather than using `math.sqrt()` one at a time.

````python
%%time 
numbers_arr = np.arange(50_000_000)
numbers_sqrt_np = np.sqrt(numbers_arr)
````

This took only ~150 ms (0.15s) rather than 1.89s! This is a relatively crude test, and also included the _generation_ of 50 million numbers. In practice, speed ups can be even faster in a lot of situations.

### Numpy array characteristics

We can inspect a number of potential characteristics of our numpy array, such as the number of dimensions (`ndim`) and the length of those dimensions (`shape`):

````python
a = np.array([1, 2, 3, 4])

print(a.ndim)
print(a.shape)
````

Our array has one dimension, which has a length four.

This doesn't appear particularly useful, but numpy arrays can be $n$-dimensional - for instance, here is a 2D array:

````python
b = np.array([[1, 2, 3], [4, 5, 6]])

print(b)

print('Number of dimensions:', b.ndim)
print('Shape:', b.shape)
````

Our 2-D array has arrays of length 2 and 3.

We can also explore the data type of a numpy array, through `dtype`:

````python
b.dtype
````

This shows that is a 64-bit integer. This means the maximum value this data type could hold is 2<sup>64</sup>, which is --

````python
2**64
````

-- quite large! If we had a very very large array and were worried about file sizes, we could instead select a smaller data type, such as `int32` or even `int8`. `int8` stores 2<sup>8</sup>, or 256, values (this value might seem familiar from various interactions with computers over the years!). Many old digital satellite images are stored in 8-bit colour scale, with a single value between 0 and 255. So there's no point storing old satellite data with more precision than this - you're just wasting file space!

You can change data types using `astype`. Let's change this to a 32-bit float.

````python
c = b.astype(np.float32)  # 'astype' will also accept a suitable string representation of the data type, e.g., 'float32', 'int64', etc.

print(c)

print(c.dtype)
````

### Generating numpy arrays

We know about the existing `range()` function, but numpy has its own functions - one of which we used above. These have some nice additional functionality, such as stepping within the `np.arange(start, stop, step)` function:

````python
d = np.arange(1, 10, 2)  # Start at 1, go up to (but not including) 10, step by 2

d
````

`linspace(start, stop, num)` allows us to create a range of `num` evenly spaced values between `start` and `stop`:

````python
e = np.linspace(0, 10, 21)  # Start at 0, go up to (and include) 10, with 21 evenly spaced points (i.e. every 0.5)

e
````

### Constants and Functions

As under `math`, `numpy` has a standard range of constants and functions available to you. Here are some examples of things you can do, mixed with approaches we learned last week: see if you can understand what's going on.

````python
print(np.pi)
````

````python
x = np.linspace(0, 2 * np.pi, 10)  # Create an array of 10 evenly spaced values from 0 to 2π

y = np.sin(x)  # Calculate the sine of each value in x


for x_i, y_i in zip(x, y):  # "zip" combines the two arrays into pairs of (x, y) values

    # Print the x and sin(x) values, formatted to 2 decimal places
    print(f"x: {x_i:.2f}, sin(x): {y_i:.2f}")  # adding ":.2f" to an f-string formats the number to 2 decimal places
````

````python
# Round y to three decimal places
y_rounded = np.round(y, 3)

print("y (original):", y)
print("y (rounded to 3 decimal places):", y_rounded)
````

````python
# Convert between radians and degrees

x_deg = np.rad2deg(x)  # x ranges from 0 to 2π, so after converting from radians to degrees, x_deg should range from 0 to 360 degrees

x_deg
````

````python
# Calculate the sum, and cumulative sum, of a sequence of numbers

z = np.arange(1, 11)  # Create an array of numbers from 1 to 10

z_sum = np.sum(z)  # Calculate the sum of the elements in z

z_cumsum = np.cumsum(z)  # Calculate the cumulative sum of the elements in z

print(f"z: {z}")
print(f"Sum of z: {z_sum}")
print(f"Cumulative sum of z: {z_cumsum}")
````

````python
# Calcualte statistics of an array

w = np.random.rand(100)  # Create an array of 1000 random numbers between 0 and 1

w_mean = np.mean(w)  # Mean

w_std = np.std(w)  # Standard deviation

w_median = np.median(w)  # Median

print(f"Mean of w: {w_mean}")
print(f"Standard deviation of w: {w_std}")
print(f"Median of w: {w_median}")
````

````python
# Use some conditional logic to filter an array

w = np.random.rand(10)  # Create an array of 100 random numbers between 0 and 1

w_mask = w > 0.5  # Select only the elements of w that are greater than 0.5

print("w_mask shows True where the condition is met and False where it is not:", w_mask)

w_filtered = w[w > 0.5]  # Use the mask to filter the array

print("We can filter to show only elements of w that are greater than 0.5:", w_filtered)
````

### Indexing and Slicing

As with lists, we can use indexing to pull out individual elements. This can be more confusing for multidimensional arrays. Let's create a 2-dimensional array, using the `reshape` function:

````python
f = np.arange(12).reshape(3, 4)  # Create a 3x4 array
f
````

Recall that Python indexing starts at `0`: if we select the first element, we will get:

````python
f[0]
````

Looking at how our array is structured above, this makes some sense: our n-dimensional array is basically an 1-D array made up of three arrays. So `f[0]` brings out this first array. 

We can further pull out a specific item in this array:

````python
f[0, 2]  # Select the first row and the third column
````

````python
f[-1, -3]  # Select the last row and the third-to-last column
````

Play around with this multi-dimensional indexing until you feel comfortabl ewith what is going on.

We can also extract ranges of data using slices. We did this previously to lists, but again, we can do _multidimensional_ slicing by repeating our instruction for the row, and then the column:

````python
f[0:2, 0:2]  # Select the first two rows and the first two columns
````

````python
f[0:2, ::]  # Select the first two rows and all columns (note the special "::" syntax)
````

Several functions allow us to provide an axis, so that operations can be performed only in one direction. For instance, instead of the default `sum`, we can direct the sum along specific axes. Remember that, within `numpy` convention, axis `0` is along-column, and axis `1` is along-row.

````python
print(f.sum())  # Calculate the sum of all elements in the array f

print(f.sum(axis=0))  # Calculate the sum of each column

print(f.sum(axis=1))  # Calculate the sum of each row
````
