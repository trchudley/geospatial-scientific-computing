# Plotting and `matplotlib`

## Introduction

At a certain point, we should stop looking at large printouts of numbers and start being able to plot some data! There are a few competing plotting libraries in Python, but the one Python learners generally start with (and also the most popular, and still the best, in my opinion, for producing static plots for reports, articles, etc.) is [matplotlib](https://matplotlib.org/). 

"Matplotlib" is short for "MATLAB Plotting Library", a vestigial reference to the days when the package was trying to emulate the MATLAB style of plotting. It has since diverged in many ways from MATLAB, and offers a robust variety of output formats, chart types, and interactive options.

## Simple Plotting

We can import Matplotlib's plotting library, the `pyplot` interface. Traditionally, we import it to the variable name `plt`. Whilst we're here, we will also import `numpy`. We will do this both in the same cell, at the top of the code. It is good practice to import all of your external packages/functions at the top of the code: it means that any reader can tell, at a glance, what has been requried to be installed and is going to be used within a session.

````python
import matplotlib.pyplot as plt
import numpy as np
````

Let's generate some test data, in the form of numpy arrays:

````python
x = np.linspace(0, 10, 20)  # Create an array of 20 evenly spaced values from 0 to 10

x_squared = x ** 2
x_cubed = x ** 3
x_sin = np.sin(x)
x_cos = np.cos(x)
````

### Our first figure

Now, we can create our first plot. A matplotlib figure has two parts: a `Figure`, which is the entire 'plot', which can contain one or more `Axes`, which have a variety of editable properties. Let's show a really simple example of how we can plot a line graph: we will create our matplotlib figure/axis, and then use the `scatter` function of the Axis object to plot our `x` data against `x_squared`:

````python
fig, ax = plt.subplots()  # Create a figure and axis object for plotting

ax.scatter(x, x_squared)

plt.show()
````

It's as simple as that! We can add a number of different elements to the plot, such as axis labels, titles, or a background grid. Below, we will also use the `plot` rather than `scatter` function: `plot` is the line plot function.

````python
fig, ax = plt.subplots()  # Create a figure and axis object for plotting

ax.plot(x, x_squared)

ax.set_xlabel('x')  # Set the x-axis label
ax.set_ylabel('x squared')  # Set the y-axis label

ax.grid() # Add a grid to the plot

ax.set_title('Line plot of x vs x squared')  # Set the title of the plot

plt.show()
````

We can show multiple data within the same plot, and add 'labels' and a legend to describe them:

````python
fig, ax = plt.subplots()

ax.plot(x, x_squared, label='x ** 2')
ax.plot(x, x_cubed, label='x ** 3')

ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_title('Raising x to different powers')

ax.legend()  # Add a legend to the plot, which draws from the 
             # `label` arguments in the `plot` function

plt.show()
````

We’re not restricted to the default look for plot elements. 

Most plot elements have style attributes, such as `linestyle` and `color`, that can be modified to customize the look of a plot. The color attribute can accept a wide array of color options in [a variety of named `colors` like `red` or `blue`](https://matplotlib.org/stable/gallery/color/named_colors.html). Here, we use some different shades of red taken from the modern "Tableau" colorset in Matplotlib, which can by prepending typical colours with `tab` -- e.g. using the `tab:red` option for the color attribute.

We can also use `xlim` and `ylim` commands to clip outputs to a pre-determined space range.

````python
fig, ax = plt.subplots()

ax.plot(x, x_sin, label='sin(x)', color='tab:green', linestyle='--')  # dashed linestyle
ax.plot(x, x_cos, label='cos(x)', color='tab:purple', linestyle=':')  # dotted linestyle

# These commands can be set in any arbritrary order.

ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_title('Sine and Cosine functions')

ax.legend()

ax.set_xlim(0, 7.8)
ax.set_ylim(-1.5, 2.5)
````

There are also custom modifications you can request for the dotted scatterplot:

````python
fig, ax = plt.subplots()

ax.scatter(x, x_sin, label='sin(x)', color='tab:red', marker='x', s=20)  # 's' is the size of the marker
ax.scatter(x, x_cos, label='cos(x)', color='tab:olive', marker='o', s=50)

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Scatterplot of sin(x) and cos(x)')
ax.legend()

plt.show()
````

## Other Plot Styles

### Bar plot

Using `ax.bar()`, we can also create a simple bar plot. Note that here, our `x` values are a list of string labels, rather than values.

````python
# Example bar plot

labels = ['A', 'B', 'C', 'D', 'E']
values = [5, 7, 3, 8, 4]

fig, ax = plt.subplots()

ax.bar(labels, values, color='tab:blue')

ax.set_xlabel('Categories')
ax.set_ylabel('Values')
ax.set_title('Bar plot example')

plt.show()
````

### Histograms

We can also present a slightly more complicated histogram, showing the frequency distribution of a range of values. Note that we can request by how much our data is split using the `bins` parameter: I have requested 30 bins here. I have also explicitly set a `black` `edgecolor`, so you can see where the bins end.

````python
data = np.random.normal(loc=0, scale=1, size=1000)  # Generate 1000 random numbers from a normal distribution

fig, ax = plt.subplots()

ax.hist(data, bins=30, color='tab:blue', edgecolor='black')  # 'bins' specifies the number of bins in the histogram

ax.set_xlabel('Value')
ax.set_ylabel('Frequency')
ax.set_title('Histogram of normally distributed data')

plt.show()
````

Note that, by default, the `hist` function will shows a frequency plot. A frequency plot is a histogram where the y-axis shows the frequency of each bin, rather than the count. We can change this to a count of values within each bin using the `density=False` function:

````python
# Second histogram example, this time showing how to get counts rather than frequency

data = np.random.normal(loc=0, scale=1, size=1000)

fig, ax = plt.subplots()

ax.hist(data, bins=30, color='tab:blue', edgecolor='black', density=False)  # 'density=False' means we want counts

ax.set_xlabel('Value')
ax.set_ylabel('Counts')
ax.set_title('Histogram of normally distributed data (counts)')

plt.show()
````

If you're not happy with the bin edges as described, you can supply custom bin edges to the `bins` parameter:

````python
# Third histogram example, showing how to set custom bin edges

data = np.random.normal(loc=0, scale=1, size=1000)

fig, ax = plt.subplots()

# Custom bin edges, every 0.4 between -3 and 3
bin_edges = np.arange(-3, 3.2, 0.4)

ax.hist(data, bins=bin_edges, color='tab:blue', edgecolor='black', density=False)

ax.set_xlabel('Value')
ax.set_ylabel('Counts')
ax.set_title('Histogram with custom bin edges')

plt.show()
````

### Adding colours to a plot to show a third variable

Sometimes, you might be interested in showing a third dimension of data within your plots. You can use the `c` variable not just to supply a single colour (e.g. `red`), but also a third axis of data for your points to be coloured by. For example, maybe you might want to show the following:

 - `x` is the longitude of a set of points, `y` is the latitude, and `z` is some measurement at those points (e.g., temperature, elevation, etc.)
 - `x` is the age of a set of individuals, `y` is their income, and `z` is their level of education
 - `x` is the time of day, `y` is the number of customers in a store, and `z` is the average amount spent per customer


You can do this as follows:

````python
x = np.random.rand(100)  # 100 random x values
y = np.random.rand(100)  # 100 random y values
z = np.random.rand(100)  # 100 random values for color mapping

# Some example of what the x, y, and z arrays might actually be in real applications:

fig, ax = plt.subplots()

scatter = ax.scatter(x, y, c=z, cmap='viridis')  # 'c' specifies the color values, 'cmap' specifies the colormap

plt.colorbar(scatter, label='A random value between 0 and 1')  # Add a colorbar to show the mapping between colors and values

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Scatter plot with color mapping')

plt.show()
````

## Subplots

Finally, it is all very well showing individual plots, but sometimes you might want to show a multiple plots together. Here, we can begin to explore the `plt.subplots()` function, that, up until now, we have been calling without providing any arguments.

Within `plt.subplots()`, we can set the number of rows and columns (e.g. `ncols=2, nrows=3`) as well as the overall figure size (`figsize=(10, 4)`). Unfortunately, `figsize` is in inches rather than centimetres or millimetres - but it's useful to figure out what final size you might want your plot, as things such as text size will be scaled within this space.

Another slightly annoying but necessary additional parameter is the use of a `layout` parameter. Without this, you might find that your plots begin to overlap and look awkward - by setting `layout='constrained'`, the plots will be arranged nicely.

````python
x = np.linspace(0, 2 * np.pi, 100)  # 100 points from 0 to 2π
a = np.sin(x)  # Sine values
b = np.cos(x)  # Cosine values

fig, axes = plt.subplots(
    ncols=2,                # Create a figure with 2 columns of subplots
    nrows=1,                # Create a figure with 1 row of subplots
    figsize=(10, 4),        # Create a figure with 1 row and 2 columns of subplots
    layout='constrained'    # Use a 'constrained' layout to arrange subplots nicely and avoid overlap of labels and titles
)  

ax = axes[0]  # First subplot

ax.plot(x, a, label='sin(x)', color='tab:blue')
ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
ax.set_title('Sine function')

ax = axes[1]  # Second subplot
ax.plot(x, b, label='cos(x)', color='tab:orange')
ax.set_xlabel('x')
ax.set_ylabel('cos(x)')
ax.set_title('Cosine function')

# Title for whole figure
plt.suptitle('Sine and Cosine functions')

plt.show()

````

Note that, in this setup, we go from `fig, ax = plt.subplots()` to `fig, axes = plt.subplots(ncols=2)`. We are now returning a `tuple` of axes, rather than a single axis. As a result, when we go to select a single axis in the code, we set `ax = axes[0]`, selecting the axis we want via indexing. For 2-D grids of plots (see below), you will need to select a 2D index - e.g. `ax = axes[0, 0]` for the top left.

When you have a lot of plots showing similar things, you can try and use the `sharex` and `sharey` arguments to ensure that they all occupy the same space, which also means you only have to set the `x` and `y` axis labels on the appropriate edge of rows and columns.

````python
x_col1 = np.linspace(0, 10, 100)  # 100 points from 0 to 10
x_col2 = np.linspace(0, 10, 7)  # 100 points from 0 to 10

y_col1row1 = np.sin(x_col1)
y_col2row1 = np.sin(x_col2)
y_col1row2 = np.sin(x_col1) * 2
y_col2row2 = np.sin(x_col2) * 2
y_col1row3 = np.sin(x_col1) * 4
y_col2row3 = np.sin(x_col2) * 4
y_col1row4 = np.sin(x_col1) * 6
y_col2row4 = np.sin(x_col2) * 6

fig, axes = plt.subplots(
    ncols=2,                # Create a figure with 2 columns of subplots
    nrows=4,                # Create a figure with 4 rows of subplots
    figsize=(6, 8),       # Create a figure with 1 row and 2 columns of subplots
    layout='constrained',   # Use a 'constrained' layout to arrange subplots nicely and avoid overlap of labels and titles
    sharex=True,            # Share the x-axis across all subplots
    sharey=True             # Share the y-axis across all subplots
)

plt.suptitle('Sine function with different multipliers and resolutions')  # Title for whole figure

ax = axes[0, 0]  # First subplot (row 0, column 0)
ax.plot(x_col1, y_col1row1, label='sin(x)', color='tab:blue')
ax.set_ylabel('sin(x)')  # Set y-axis label for the first row
ax.set_title("High resolution x values")  # Set title for the first column

ax = axes[0, 1]  # Second subplot (row 0, column 1)
ax.plot(x_col2, y_col2row1, label='sin(x + 1)', color='tab:orange')
ax.set_title("Low resolution x values")  # Set title for the second column

ax = axes[1, 0]  # Third subplot (row 1, column 0)
ax.plot(x_col1, y_col1row2, label='sin(x + 2)', color='tab:green')
ax.set_ylabel('sin(x) * 2')  # Set y-axis label

ax = axes[1, 1]  # Fourth subplot (row 1, column 1)
ax.plot(x_col2, y_col2row2, label='sin(x + 3)', color='tab:red')

ax = axes[2, 0]  # Fifth subplot (row 2, column 0)
ax.plot(x_col1, y_col1row3, label='sin(x + 4)', color='tab:purple')
ax.set_ylabel('sin(x) * 3')  # Set y-axis label

ax = axes[2, 1]  # Sixth subplot (row 2, column 1)
ax.plot(x_col2, y_col2row3, label='sin(x + 5)', color='tab:brown')

ax = axes[3, 0]  # Seventh subplot (row 3, column 0)
ax.plot(x_col1, y_col1row4, label='sin(x + 6)', color='tab:pink')
ax.set_ylabel('sin(x) * 4')  # Set y-axis label
ax.set_xlabel('x (high resolution)')  # Set x-axis label for the first row

ax = axes[3, 1]  # Eighth subplot (row 3, column 1)
ax.plot(x_col2, y_col2row4, label='sin(x + 7)', color='tab:gray')
ax.set_xlabel('x (low resolution)')  # Set x-axis label for the last row
````

## Saving plots

When you have a nice plot you want to save, there are options available for you. The core of this is `plt.savefig()` function.

You can save the figure in different formats, such as PNG, PDF, SVG, etc. For 'raster' images (jpg, png, etc.), the 'dpi' parameter specifies the resolution of the saved image in dots per inch (DPI). A higher DPI value results in a higher resolution image, but will also take up more space. For plots made up of exclusively lines and shapes, like we have here, you may wish to consider a vector format (pdf, eps, etc.), which will be much less space-intensive whilst retaining good quality even when zoomed in.


````python
# Example of how to save a figure

x = np.linspace(0, 10, 100)
y = np.sin(x)

fig, ax = plt.subplots()

ax.plot(x, y, label='sin(x)', color='tab:blue')

ax.set_xlabel('x')
ax.set_ylabel('sin(x)')
ax.set_title('Sine function')

plt.savefig('sine_function_plot.png', dpi=300)  # Save the figure as a PNG file with 300 dpi resolution
plt.savefig('sine_function_plot.pdf')  # Save the figure as a PDF file

plt.show()
````

By default, the figure will be saved in the current working directory. You can specify a different path if you want to save it elsewhere - for example, `plt.savefig('C:/Users/YourUsername/Documents/sine_function_plot.png')` on Windows or `plt.savefig('/home/yourusername/Documents/sine_function_plot.png')` on Linux/Mac.

Try opening the `png` and `pdf` figures and zoom in - see how a pixel-based raster (best for storing images) differs from a geometry-based vector image (best for storing shapes). Keep this difference in mind - raster and vector data will come back in a big way when we start thinking about geospatial data!
