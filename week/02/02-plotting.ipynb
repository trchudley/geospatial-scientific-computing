## Introduction

At a certain point, we need to stop looking at large printouts of numbers and start being able to plot some data! There are a few competing plotting libraries in Python, but the one Python learners generally start with (and also the most popular, and still the best, in my opinion, for producing static plots for reports, articles, etc.) is [matplotlib](https://matplotlib.org/). "Matplotlib" is short for "MATLAB Plotting Library", a vestigial reference to the days when it was trying to emulate the MATLAB style of plotting. It has since diverged in many ways from MATLAB, and offers a robust variety of output formats, chart types, and interactive options.

## Simple Plotting

We can the Matplotlib's plotting library, the `pyplot` interface. Traditionally, we import it to the variable name `plt`. Whilst we're here, we will also import `numpy`. We will do this both in the same cell, at the top of the code. This is a common way of doing things: it means that any reader can tell, at a glance, what has been imported and needs to be used within a session.

````python
import matplotlib.pyplot as plt
import numpy as np
````

Let's generate some test data, in the form of numpy arrays:

````python
x = np.linspace(0, 10, 20)  # Create an array of 20 evenly spaced values from 0 to 10

x_squared = x ** 2
x_cubed = x ** 3
x_sin = np.sin(x)
x_cos = np.cos(x)
````

### Our first figure

Now, we can create our first plot. A matplotlib figure has two parts: a `Figure`, which is the entire 'plot', which can contain one or more `Axes`, which have a variety of editable properties. Let's show a really simple example of how we can plot a line graph: we will create out matplotlib figure/axis, and then use the `scatter` function of the Axis object to plot our `x` data against `x_squared`:

````python
fig, ax = plt.subplots()  # Create a figure and axis object for plotting

ax.scatter(x, x_squared)

plt.show()
````

It's as simple as that! We can add a number of different elements to the plot, such as axis labels, titles, or a background grid. Below, we will also use the `plot` rather than `scatter` function: `plot` shows a line plot.

````python
fig, ax = plt.subplots()  # Create a figure and axis object for plotting

ax.plot(x, x_squared)

ax.set_xlabel('x')  # Set the x-axis label
ax.set_ylabel('x squared')  # Set the y-axis label

ax.grid() # Add a grid to the plot

ax.set_title('Line plot of x vs x squared')  # Set the title of the plot

plt.show()
````

We can show multiple data within the same plot, and add 'labels' and a legend to describe them:

````python
fig, ax = plt.subplots()

ax.plot(x, x_squared, label='x ** 2')
ax.plot(x, x_cubed, label='x ** 3')

ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_title('Raising x to different powers')

ax.legend()  # Add a legend to the plot

plt.show()
````

We’re not restricted to the default look for plot elements. 

Most plot elements have style attributes, such as `linestyle` and `color`, that can be modified to customize the look of a plot. The color attribute can accept a wide array of color options, including keywords (named `colors`) like `red` or `blue`. Here, we use some different shades of red taken from the "Tableau" colorset in Matplotlib, which can by prepending typical colours with `tab` -- e.g. using the `tab:red` option for the color attribute.

We can also use `xlim` and `ylim` commands to clip outputs to a pre-determined space range.

````python
fig, ax = plt.subplots()

ax.plot(x, x_sin, label='sin(x)', color='tab:green', linestyle='--')  # dashed linestyle
ax.plot(x, x_cos, label='cos(x)', color='tab:purple', linestyle=':')  # dotted linestyle

# These commands can be set in any arbritrary order.

ax.set_xlabel('x')
ax.set_ylabel('y')

ax.set_title('Sine and Cosine functions')

ax.legend()

ax.set_xlim(0, 7.8)
ax.set_ylim(-1.5, 2.5)
````
