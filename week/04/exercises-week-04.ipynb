:::{note} Accessing ERA-5 data
> If you are interesting in using ERA-5 data within your final reports, you are welcome to download netCDF files from the full Climate Data Store. However, if you have no need to use large-scale gridded data, and a single point history will do - you can use the [open-meteo historical weather API](https://open-meteo.com/en/docs/historical-weather-api) to view and download (in `csv` format) hourly and monthly data at a single point.
:::
