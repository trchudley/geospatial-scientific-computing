If you would like to draw on a specific satellite image (e.g. to map a final wildfire extent), use `mapshaper.org`. If you are fine drawing on a generic satellite basemap (like e.g. Google Maps), use the slightly simpler `geojson.io`.

#### mapshaper.org

If you need to draw an outline on a specific geotiff you downloaded, you are welcome to use the very easy [mapshaper.org](https://mapshaper.org/) to create a file. The steps are as follows:

 1. Drag and drop a geotiff file into the screen (note: the RGB visualisation will default to the first three bands, so it might look a little odd, but will hopefully do for a quick sketch).
 2. Click "Draw" > "Polygons" and draw your outline.
 3. Click "Export" in the top right. Select only 'polygons', and export in your chosen file format (GeoPacakge is the best default option)

Note that the file will come encoded as WGS84 (EPSG:4326): you will need to convert this to match your raster data.

#### geojson.io

If you don't need to draw onto a specific geotiff file, and are happy drawing from a generic basemap, you can use [https://geojson.io]. The steps are as follows:

1. [Optional] Click the 'Change Basemap' symbol in the top right (the stacked squares symbol) to select a suitable background. "Standard Satellite" might be useful.
2. Click the pentagon symbol in the top left to draw a shape on the map.
3. Once you have finished drawing, click the 'Export' button in the top right and pick a file format. The default 'GeoJSON' is fine. Click 'Export' to download your file.

Note that the file will come encoded as WGS84 (EPSG:4326): you will need to convert this to match your raster data.
