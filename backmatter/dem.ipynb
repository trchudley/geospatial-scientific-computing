# Basic DEM Analysis

This is an optional section for those interested in doing some DEM analysis work. Some standard things you wish to do with DEMs, such as calculating slopes and hillshades, are very achievable but [quite awkward to recreate from first principles](https://courses.ems.psu.edu/natureofgeoinfo/print/c7_p10.html) every time you wish to calculate a variable from the raw DEM. As a result, we can take advantage of additional functions provided via [the `xarray-spatial` package](https://xarray-spatial.readthedocs.io) to automatically calculate toopgraphic derivative data with simple functions:

`xarray-spatial` aims provide fast additional capability for a variety of topographic analysis.

To ensure `xarray-spatial` is installed, enter your `conda` environment and run the command `conda install -c conda-forge rioxarray`.

'`slope`' is the steepest gradient at a point, in degrees. '`aspect`' is the compass direction that the steepest slope faces, in degrees. And '`hillshade`' is an artifical way of showing the terrain might be illuminated from a given point light source (the `azmiuth` and `angle_altitude` define the direction of this light source). The latter is useful to overlay as a semi-transparent greyscale layer (defined using the `alpha` argument) to better visualise topographic variables. An example of visualising these datasets is shown below:

```python
# plot dem, slope, and aspect in 1x3 grid, and overlay with hillshade

fig, axes = plt.subplots(nrows=3, figsize=(6, 12), layout='constrained')

ax = axes[0]
dem_reprojected.plot.imshow(ax=ax, cmap="terrain", vmin=0, vmax=230, cbar_kwargs={"label": "Elevation (m)"})
hillshade.plot.imshow(ax=ax, cmap="gray", alpha=0.5, add_colorbar=False)
ax.set_title("DEM with Hillshade")
ax.set_aspect('equal')

ax = axes[1]
slope.plot.imshow(ax=ax, cmap="viridis", vmin=0, vmax=30, cbar_kwargs={"label": "Slope (degrees)"})
hillshade.plot.imshow(ax=ax, cmap="gray", alpha=0.5, add_colorbar=False)
ax.set_title("Slope with Hillshade")
ax.set_aspect('equal')

ax = axes[2]
aspect.plot.imshow(ax=ax, cmap="twilight", vmin=0, vmax=360, cbar_kwargs={"label": "Aspect (degrees)"})
hillshade.plot.imshow(ax=ax, cmap="gray", alpha=0.5, add_colorbar=False)
ax.set_title("Aspect with Hillshade")
ax.set_aspect('equal')

plt.show()
```
